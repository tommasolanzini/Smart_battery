```python
import pandas as pd

import pandas as pd
from pathlib import Path

# ---- CONFIGURA QUI ----
files = [
    r'.\Battery Testing\test_3\1.csv',
    r'.\Battery Testing\test_3\2.csv',
    r'.\Battery Testing\test_3\3.csv'
]

output_file = r'.\Battery Testing\test_3.csv'
# ------------------------

dfs = []
time_offset = 0

for i, f in enumerate(files):
    df = pd.read_csv(f)

    # Se non è il primo file, evita duplicazione del primo campione
    if i > 0:
        df = df.iloc[1:].copy()

    # Trasla il tempo
    df['time_s'] = df['time_s'] + time_offset

    # Aggiorna offset
    time_offset = df['time_s'].iloc[-1]

    dfs.append(df)

# Concatenazione finale
merged = pd.concat(dfs, ignore_index=True)

# Salvataggio CSV unico
merged.to_csv(output_file, index=False)

print("CSV unificato salvato in:")
print(output_file)
```
